# 4. BUILD DIM_DATE

```python
from pyspark.sql.functions import col, year, month, dayofmonth, quarter

df_silver = spark.read.table('retail_project_dev.silver.sales_clean')

dim_date = (
    df_silver.select(col('TransactionDate').cast('date').alias('DateKey'))
    .dropDuplicates(['DateKey'])
    .withColumn('Day',dayofmonth(col('DateKey')))
    .withColumn('Month',month(col('DateKey')))
    .withColumn('Year',year(col('DateKey')))
    .withColumn('Quarter',quarter(col('DateKey')))

) 

dim_date.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("retail_project_dev.gold.dim_date")
```
