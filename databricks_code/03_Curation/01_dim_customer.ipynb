# 1. BUILD DIM_CUSTOMER

```python
df_silver = spark.read.table('retail_project_dev.silver.sales_clean')

dim_customer = df_silver.select('customerID', 'customerage', 'gender', 'CustomerLoyaltyLevel').dropDuplicates(["CustomerID"]).fillna({"customerage": -1})

dim_customer.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("retail_project_dev.gold.dim_customer") 
```
