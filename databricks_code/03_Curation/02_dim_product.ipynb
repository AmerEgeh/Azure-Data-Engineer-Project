# 2. BUILD DIM_PRODUCT

```python
df_silver = spark.read.table('retail_project_dev.silver.sales_clean')

dim_product = df_silver.select('ProductID', 'ProductName', 'ProductCategory', 'SubCategory', 'Brand').dropDuplicates(["ProductID"])

dim_product.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("retail_project_dev.gold.dim_product")
```
