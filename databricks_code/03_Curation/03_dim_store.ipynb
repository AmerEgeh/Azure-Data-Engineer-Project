# 3. BUILD DIM_STORE

```python
df_silver = spark.read.table('retail_project_dev.silver.sales_clean')

dim_store = df_silver.select('storeID', 'storelocation', 'storeregion').dropDuplicates(["storeID"])

dim_store.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("retail_project_dev.gold.dim_store")
```
